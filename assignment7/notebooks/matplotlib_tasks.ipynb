```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Name": ["Anna", "John", "Ali", "Maria", "Tom"],
    "Age": [19, 21, 22, 20, 23],
    "Score": [78, 85, 90, 88, 70]
}

df = pd.DataFrame(data)
```

```python
plt.plot(df["Name"], df["Score"])
plt.xlabel("Students")
plt.ylabel("Scores")
plt.title("Student Scores Line Plot")
plt.show()
```

```python
plt.bar(df["Name"], df["Score"])
plt.xlabel("Students")
plt.ylabel("Scores")
plt.title("Student Scores Bar Chart")
plt.show()
```
